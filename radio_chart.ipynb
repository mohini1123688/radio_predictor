```python
import requests #for HTTP request
from bs4 import BeautifulSoup #for HTML parsing
```

```python
# Sent during request
headers = {
    'user-agent': 'Student/ I am studying data science/ WashU in St Louis'
}

# These are URL parameters that come after the ? in URL
params = {
}

response = requests.get('https://kworb.net/radio/', params=params, headers=headers)
```

```python
print( response.status_code )
```

```python
print( response.text )
```

```python
# Parse response with HTML parser
soup = BeautifulSoup(response.text)
```

```python
print( soup.title )
```

```python
print(soup.table)
```

```python
song_number = 0
for tag in soup.find_all('td', class_ = "mp text"):
    if song_number < 100:
        print(tag.text)
        song_number += 1
```

```python
import csv
```

```python
song_number = 0
for tag in soup.find_all('tr'):
    if song_number < 100:
        print(tag.text)
        song_number += 1
```

```python
radio_file = "/Users/mohinimahajan/Documents/CSE3104_Project/radio_chart.csv"
with open(radio_file, 'w') as file:
    writer = csv.writer(file)
    for tag in soup.find_all('tr'):
        writer.writerow([tag.text])
```
